```python
PROMPT = "A red fox running through fresh snow in a pine forest at sunrise"
SEED = 42
DTYPE = "auto"        # auto | fp16 | fp32
IMAGE = None           # e.g. '/kaggle/input/my-photo/photo.jpg' to animate your own picture
INTERPOLATE = 2        # 1 = off, 2 = 97 smooth frames (RIFE)
```

```python
import shlex, subprocess
cmd = (f"python scripts/inference_neodragon.py --prompt {shlex.quote(PROMPT)} --seed {SEED} "
       f"--model_dtype {DTYPE} --tiny_first_frame_vae --interpolate {INTERPOLATE} --output_video_folder output")
if IMAGE: cmd += f" --image {shlex.quote(IMAGE)}"
print(cmd)
!{cmd}
```
